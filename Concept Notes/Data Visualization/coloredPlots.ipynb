```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
churn = pd.read_csv("Datasets/Churn_Modelling.csv")
```

```python
churn
```

```python
#Using colors as 3rd dimension in Pandas. Providing c = "Gender" directly will throw an error due to textual categorical data
churn["GenderColors"] = churn["Gender"].apply(lambda x: "blue" if x == "Male" else "red")
churn.plot.scatter(x="Age", y="CreditScore", c="GenderColors")     
```

```python
#Using colors as third dimension in Matplotlib (not ideal)
for g in ["Male", "Female"]:
    c = "blue" if g == "Male" else "red"
    gChurn = churn[churn["Gender"] == g]
    plt.scatter(x=gChurn["Age"], y=gChurn["CreditScore"], color=c)
```

```python
#Using colors as third dimension in Seaborn (ideal)
sns.scatterplot(data=churn, x="Age",y="CreditScore", hue="Gender")
```

```python
#What if a continuous parameter is to be hued? It will perform binning but differentiation still will be tough
sns.scatterplot(data=churn, x="Age",y="CreditScore", hue="EstimatedSalary")
```

```python
#Pandas can handle such cases better
churn.plot.scatter(x="Age", y="CreditScore", c="EstimatedSalary")  
```

```python
#Matplotlib also gives similar results but won't give labels and/or colorbar by default
col = plt.scatter(x=churn["Age"], y=churn["CreditScore"], c=churn["EstimatedSalary"])
plt.colorbar(col)
```
