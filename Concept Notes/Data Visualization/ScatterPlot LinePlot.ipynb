```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
churn = pd.read_csv("Datasets/Churn_Modelling.csv")
```

```python
churn.info()
```

```python
churn.sample(1000).plot.scatter(x="Age", y="CreditScore")  #Using Pandas. sample(1000) reduces points for clarity
```

```python
plt.scatter(x=churn["Age"], y=churn["CreditScore"])
plt.xlabel("Age")
plt.ylabel("Credit Score")
```

```python
sns.scatterplot(data=churn, x="Age", y="CreditScore")
```

```python
#average credit score and salary by age
churn_age_cs_sal = churn.groupby('Age')[['CreditScore', 'EstimatedSalary']].mean().reset_index()
```

```python
churn_age_cs_sal
```

```python
churn_age_cs_sal.plot.line(x = 'Age', y = 'CreditScore')         #Using Pandas to plot line
```

```python
plt.plot(churn_age_cs_sal["Age"],churn_age_cs_sal["CreditScore"])        #Using Matplotlib to plot line
```

```python
sns.lineplot(data=churn_age_cs_sal, x="Age", y="CreditScore")          #Using seaborn to plot line
```
