```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
churn = pd.read_csv("Datasets/Churn_Modelling.csv")
```

```python
churn.info()
```

```python
churn.sample(1000).plot.scatter(x="Age", y="CreditScore")  #Using Pandas. sample(1000) reduces points for clarity
```

```python
plt.scatter(x=churn["Age"], y=churn["CreditScore"])
plt.xlabel("Age")
plt.ylabel("Credit Score")
```

```python
sns.scatterplot(data=churn, x="Age", y="CreditScore")
```

```python
#average credit score and salary by age
churn_age_cs_sal = churn.groupby('Age')[['CreditScore', 'EstimatedSalary']].mean().reset_index()
```

```python
churn_age_cs_sal
```

```python
churn_age_cs_sal.plot.line(x = 'Age', y = 'CreditScore')         #Using Pandas to plot line
```

```python
plt.plot(churn_age_cs_sal["Age"],churn_age_cs_sal["CreditScore"])        #Using Matplotlib to plot line
```

```python
sns.lineplot(data=churn_age_cs_sal, x="Age", y="CreditScore")          #Using seaborn to plot line
```

```python
churnSalGen = churn.groupby("Gender")[["EstimatedSalary","CreditScore"]].mean().reset_index()
```

```python
# bar plot using Pandas, use ploth method to plot horizontally
churnSalGen.plot.bar(x="Gender", y="EstimatedSalary")
```

```python
# Bar plot using matplotlib, use barh method to plot horizontally
plt.bar(churnSalGen["Gender"], churnSalGen["EstimatedSalary"])
```

```python
#Barplot using seaborn, swap axes to plot horizontally
sns.barplot(data=churnSalGen, x="Gender", y="EstimatedSalary")
```

```python
#histogram using pandas, gender is a category => No binning required
churn["Gender"].hist()
```

```python
#Pandas will need an additional statement with method value_counts() to give output number
churn["Gender"].value_counts()
```

```python
#histogram using matplotlib. Plot as well as Store values in variables to show output in numbers
values, x, _ = plt.hist(churn["Gender"])
```

```python
#get count values for histogram categories
values
```

```python
#Binning using Pandas
churn["EstimatedSalary"].hist()
```

```python
#Binning using matplotlib. Plot as well as Store values in variables to show output in numbers
values, x, _ = plt.hist(churn["EstimatedSalary"])
```

```python
#Use variable x to see binning distribution
x
```

```python
#use values to get count fot each bin
values
```

```python
#default bins for Pandas and matplotlib = 10. pass parameter for bins to change it.
churn["EstimatedSalary"].hist(bins=100)
```

```python
values, x, _ = plt.hist(churn["EstimatedSalary"], bins=100)
```
